```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

sns.set_theme(style="ticks", palette="pastel")
tips = sns.load_dataset("tips")

df = tips.copy()
df['tip_pct'] = (df['tip'] / df['total_bill']) * 100

print("데이터셋 로드 완료:")
display(df.head())
```

```python
print("=== [1-1] 데이터 기본 정보 (info) ===")
df.info()

print("\n=== [1-2] 수치형 변수 기초통계량 (describe) ===")
display(df.describe())

print("\n=== [1-3] 범주형 변수 기초통계량 ===")
display(df.describe(include=['category']))

print("\n=== [1-4] 요일(day) 및 성별(sex) 범주별 데이터 개수 ===")
print("- 요일별 방문 수:\n", df['day'].value_counts())
print("\n- 성별 손님 수:\n", df['sex'].value_counts())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.histplot(data=df, x="total_bill", kde=True, ax=axes[0], color="skyblue")
axes[0].set_title("Total Bill Distribution (Histplot)")

sns.countplot(data=df, x="day", ax=axes[1], palette="pastel")
axes[1].set_title("Customer Count by Day (Countplot)")

plt.tight_layout()
plt.show()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.scatterplot(data=df, x="total_bill", y="tip", hue="sex", style="smoker", ax=axes[0])
axes[0].set_title("Total Bill vs Tip (by Sex & Smoker)")

sns.boxplot(x="day", y="total_bill", hue="smoker", palette=["m", "g"], data=df, ax=axes[1])
sns.despine(offset=10, trim=True)
axes[1].set_title("Total Bill by Day and Smoker Status")

plt.tight_layout()
plt.show()
```
